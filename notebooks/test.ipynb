```python
from datetime import datetime, timedelta
```

```python

start = datetime(2026,11,9)
end = datetime(2026,11,9)

delta_time=end-start
```

```python
y=delta_time.days
```

```python
if y<0:
    print(ValueError)
    
else:
    result = []
    for i in range(y+1):
        day=start+timedelta(days=i)
        result.append(day.strftime("%d %B %Y"))
```
